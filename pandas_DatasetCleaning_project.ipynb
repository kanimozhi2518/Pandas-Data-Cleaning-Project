```python
#13.Replace missing values in Payment_Status with "Unknown".
df['Payment_Status'].fillna("Unknown",inplace=True)
df
```

```python
#14 Fill missing values in Delivery_Days using median.
# First, convert 'Delivery_Days' to numeric, coercing errors to NaN
df['Delivery_Days'] = pd.to_numeric(df['Delivery_Days'], errors='coerce')
# Then, fill missing values with the median of the now numeric column
df['Delivery_Days'].fillna(df['Delivery_Days'].median(), inplace=True)
df
```

```python
#PART - 4
#17.Remove leading and trailing spaces in all string columns.
df=df.apply(lambda x: x.str.strip() if x.dtype == "object" else x)
df
```

```python
#18.Convert Payment_Status to lowercase.
df['Payment_Status']=df['Payment_Status'].str.lower()
df
```

```python
#19.Standardize values in Customer_City.
df['Customer_City']=df['Customer_City'].str.title()
df
```

```python
#20.Clean Coupon_Code (remove extra spaces, make uppercase).
df['Coupon_Code']=df['Coupon_Code'].str.strip().str.upper()
df['Coupon_Code']
```

```python
#21.Replace "unknown" values with NaN.
import numpy as np
df.replace({'unknown':np.nan})
df
```

```python
#22.Replace "NONE" and "none" with NaN.
df.replace({'NONE':np.nan,'none':np.nan})
df
```

```python
#PART - 5
#23.Convert Delivery_Days to numeric (remove "days").
df['Delivery_Days']=df['Delivery_Days'].astype(int)
df['Delivery_Days'].fillna(df['Delivery_Days'].median(), inplace=True)
df
```

```python
#24.Convert Discount_Percent to numeric (remove %).
df['Discount_Percent']=df['Discount_Percent'].replace('%', '', regex=True)
df['Discount_Percent']=pd.to_numeric(df['Discount_Percent'],errors='coerce')
df['Discount_Percent'] = df['Discount_Percent'].fillna(df['Discount_Percent'].mode()[0])
df['Discount_Percent']=df['Discount_Percent'].astype(int)
df
```

```python
#25.Convert Support_Calls to numeric (remove "calls").
df['Support_Calls']=df['Support_Calls'].replace('Call','',regex=True)
df['Support_Calls']=pd.to_numeric(df['Support_Calls'],errors='coerce')
df['Support_Calls']=df['Support_Calls'].fillna(df['Support_Calls'].mode()[0])
df['Support_Calls']=df['Support_Calls'].astype(int)
df
```

```python
#26.Convert Customer_Rating to numeric.
df['Customer_Rating']=pd.to_numeric(df['Customer_Rating'])
df
```
